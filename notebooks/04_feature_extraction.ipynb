## Ekstrakcja cech z segmentów EKG

Każda obserwacja reprezentuje pojedyncze uderzenie serca wycięte wokół wykrytego R-peaku. 
Segment składa się z 280 kolejnych próbek sygnału EKG. Zamiast wykorzystywania wszystkich próbek 
bezpośrednio jako cech modelu, z każdego segmentu można wyznaczyć zestaw cech opisujących jego 
amplitudę, zmienność, energię oraz kształt.

### 1. Średnia (Mean)

Średnia wartość amplitudy sygnału w analizowanym segmencie:

$$
\mu = \frac{1}{N}\sum_{i=1}^{N}x_i
$$

gdzie:
- $x_i$ – wartość i-tej próbki sygnału,
- $N$ – liczba próbek segmentu.

Cecha opisuje średni poziom sygnału w obrębie pojedynczego uderzenia.

---

### 2. Odchylenie standardowe (Standard Deviation)

Miara zmienności amplitudy sygnału:

$$
\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2}
$$

Duża wartość odchylenia standardowego oznacza większą zmienność amplitudy w obrębie segmentu.

---

### 3. Wartość minimalna (Minimum)

Najmniejsza wartość amplitudy występująca w segmencie:

$$
x_{min} = \min(x_1,x_2,\ldots,x_N)
$$

Może opisywać między innymi głębokość ujemnych wychyleń sygnału.

---

### 4. Wartość maksymalna (Maximum)

Największa wartość amplitudy występująca w segmencie:

$$
x_{max} = \max(x_1,x_2,\ldots,x_N)
$$

Pozwala opisać największe dodatnie wychylenie sygnału w analizowanym uderzeniu.

---

### 5. Rozstęp amplitudy (Peak-to-Peak Amplitude)

Różnica pomiędzy maksymalną i minimalną wartością sygnału:

$$
A_{pp} = x_{max} - x_{min}
$$

Cecha opisuje całkowity zakres zmian amplitudy w obrębie uderzenia.

---

### 6. Mediana (Median)

Wartość środkowa uporządkowanych wartości amplitudy segmentu:

$$
Med = median(x_1,x_2,\ldots,x_N)
$$

W przeciwieństwie do średniej mediana jest mniej podatna na pojedyncze ekstremalne wartości sygnału.

---

### 7. Rozstęp międzykwartylowy (Interquartile Range, IQR)

Różnica pomiędzy trzecim i pierwszym kwartylem:

$$
IQR = Q_3-Q_1
$$

Cecha opisuje rozproszenie środkowych 50% wartości sygnału i jest mniej podatna na wartości odstające niż odchylenie standardowe.

---

### 8. Energia sygnału (Signal Energy)

Suma kwadratów amplitud wszystkich próbek segmentu:

$$
E = \sum_{i=1}^{N}x_i^2
$$

Energia pozwala opisać łączną wielkość zmian amplitudy występujących w obrębie pojedynczego uderzenia.

---

### 9. RMS (Root Mean Square)

Pierwiastek ze średniej wartości kwadratów amplitudy:

$$
RMS = \sqrt{\frac{1}{N}\sum_{i=1}^{N}x_i^2}
$$

RMS opisuje efektywną wielkość amplitudy sygnału, uwzględniając zarówno dodatnie, jak i ujemne wychylenia.

---

### 10. Skośność (Skewness)

Miara asymetrii rozkładu wartości amplitudy:

$$
Skewness =
\frac{1}{N}
\sum_{i=1}^{N}
\left(\frac{x_i-\mu}{\sigma}\right)^3
$$

Wartość bliska zeru wskazuje na bardziej symetryczny rozkład amplitud. Dodatnia lub ujemna skośność wskazuje na asymetrię rozkładu.

---

### 11. Kurtoza (Kurtosis)

Miara kształtu rozkładu wartości amplitudy, związana między innymi z występowaniem wartości ekstremalnych:

$$
Kurtosis =
\frac{1}{N}
\sum_{i=1}^{N}
\left(\frac{x_i-\mu}{\sigma}\right)^4
$$

Cecha może być użyteczna do wykrywania uderzeń o nietypowych, gwałtownych zmianach amplitudy.

---

### 12. Maksymalna bezwzględna amplituda

Największa bezwzględna wartość amplitudy:

$$
A_{max} = \max_i |x_i|
$$

Pozwala określić wielkość najsilniejszego wychylenia niezależnie od jego kierunku.

---

### 13. Średnia bezwzględna amplituda (Mean Absolute Amplitude)

Średnia wartość bezwzględna amplitudy:

$$
MAA = \frac{1}{N}\sum_{i=1}^{N}|x_i|
$$

Opisuje przeciętną wielkość wychylenia sygnału względem poziomu zerowego.

---

### 14. Długość linii sygnału (Signal Length / Waveform Length)

Suma bezwzględnych różnic pomiędzy kolejnymi próbkami:

$$
WL = \sum_{i=1}^{N-1}|x_{i+1}-x_i|
$$

Cecha opisuje złożoność i szybkość zmian kształtu sygnału. Segmenty zawierające gwałtowne zmiany morfologii mogą charakteryzować się większą długością linii.

---

### 15. Maksymalna zmiana pomiędzy sąsiednimi próbkami

Największa bezwzględna różnica pomiędzy kolejnymi wartościami sygnału:

$$
D_{max} = \max_i |x_{i+1}-x_i|
$$

Cecha opisuje największą lokalną szybkość zmiany amplitudy w segmencie.

---

## Zestaw cech

Dla każdego segmentu EKG można zatem utworzyć wektor cech:

| Cecha | Opis |
|---|---|
| `mean` | średnia amplituda |
| `std` | odchylenie standardowe |
| `min` | minimalna amplituda |
| `max` | maksymalna amplituda |
| `peak_to_peak` | różnica między maksimum i minimum |
| `median` | mediana amplitudy |
| `iqr` | rozstęp międzykwartylowy |
| `energy` | energia sygnału |
| `rms` | wartość skuteczna RMS |
| `skewness` | skośność rozkładu amplitudy |
| `kurtosis` | kurtoza rozkładu amplitudy |
| `max_abs` | maksymalna bezwzględna amplituda |
| `mean_abs` | średnia bezwzględna amplituda |
| `waveform_length` | suma zmian amplitudy pomiędzy kolejnymi próbkami |
| `max_diff` | największa zmiana pomiędzy sąsiednimi próbkami |

Otrzymany wektor cech może zostać wykorzystany jako reprezentacja pojedynczego uderzenia EKG na wejściu modelu One-Class SVM.

```python
import pandas as pd
import numpy as np
from scipy.stats import iqr, skew, kurtosis
```

```python
data_path = '../src/clean_data.csv'
```

```python
data = pd.read_csv(data_path, index_col=0)
```

```python
data.columns
```

```python
data.head()
```

```python
signal_columns = [f'x_{i}' for i in range(280)]
```

```python
id_columns = [
    'record_id',
    'r_peak',
    'annotation_sample',
    'type',
    'dist_from_annotation',
    'anomaly'
]

features_df = data[id_columns].copy()
```

```python
signal_values = data[signal_columns].to_numpy()

features_df['mean'] = np.mean(signal_values, axis=1)
features_df['std'] = np.std(signal_values, axis=1)
features_df['min'] = np.min(signal_values, axis=1)
features_df['max'] = np.max(signal_values, axis=1)
features_df['peak_to_peak'] = features_df['max'] - features_df['min']
features_df['median'] = np.median(signal_values, axis=1)
features_df['iqr'] = iqr(signal_values, axis=1)
features_df['energy'] = np.sum(
    signal_values ** 2, 
    axis=1
)
features_df['rms'] = np.sqrt(
    np.mean(signal_values ** 2, axis=1)
)
features_df['skew'] = skew(signal_values, axis=1)
features_df['kurtosis'] = kurtosis(signal_values, axis=1)
features_df['max_abs'] = np.max(
    np.abs(signal_values),
    axis=1
)
features_df['mean_abs'] = np.mean(
    np.abs(signal_values),
    axis=1
)
diff = np.diff(signal_values, axis=1)
features_df['waveform_length'] = np.sum(
    np.abs(diff),
    axis=1
)
features_df['max_diff'] = np.max(
    np.abs(diff),
    axis=1
)
```
