```python
import os
import wfdb
import pandas as pd
```

```python
input_path = '../data/'
output_path = '../mitbih_database/'

os.makedirs(output_path, exist_ok=True)
```

```python
record_names = [
    os.path.splitext(f)[0]
    for f in os.listdir(input_path)
    if f.endswith('.hea')
]
```

```python
record_names[0]
```

```python
for record_name in record_names:

    record_path = os.path.join(input_path, record_name)

    # =========================================================
    # 1. ECG -> CSV
    # =========================================================

    record = wfdb.rdrecord(record_path)

    ecg = pd.DataFrame(
        record.p_signal,
        columns=record.sig_name
    )

    # dodaj numer próbki
    ecg.insert(
        0,
        'sample #',
        range(len(ecg))
    )


    # zapisz CSV
    csv_path = os.path.join(
        output_path,
        f'{record_name}_record.csv'
    )

    ecg.to_csv(
        csv_path,
        index=False
    )


    # =========================================================
    # 2. Annotations -> TXT
    # =========================================================

    annotation = wfdb.rdann(
        record_path,
        'atr'
    )

    # czas anotacji
    fs = record.fs

    times = annotation.sample / fs

    # zamiana sekund na format H:MM:SS.mmm
    formatted_times = []

    for t in times:

        hours = int(t // 3600)
        minutes = int((t % 3600) // 60)
        seconds = t % 60

        formatted_times.append(
            f'{hours}:{minutes:02d}:{seconds:06.3f}'
        )


    # utworzenie tabeli
    annotations = pd.DataFrame({
        'Time': formatted_times,
        'Sample #': annotation.sample,
        'Type': annotation.symbol,
        'Sub': annotation.subtype,
        'Chan': annotation.chan,
        'Num': annotation.num,
        'Aux': annotation.aux_note
    })


    # zapis TXT
    txt_path = os.path.join(
        output_path,
        f'{record_name}_annotation.txt'
    )

    annotations.to_csv(
        txt_path,
        sep='\t',
        index=False
    )


    print(f'{record_name} -> done')
```
